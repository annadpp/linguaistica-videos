```python
# Averiguamos qué versión de NLTK y spaCy tenemos
print(f"✅ NLTK versión: {nltk.__version__}")
print(f"✅ spaCy versión: {spacy.__version__}")
```

```python
# Cargamos el modelo de español de spaCy
nlp = spacy.load("es_core_news_sm")
print("✅ Modelo de español cargado correctamente")
```

```python
# Probamos NLTK: tokenización
from nltk.tokenize import word_tokenize
frase = "Bienvenidos a linguAIstica. Espero que aprendamos mucho sobre lingüística computacional."
tokens = word_tokenize(frase, language="spanish")
print(f"🔹 Tokens con NLTK: {tokens}")
```

```python
# Probamos spaCy: análisis gramatical
doc = nlp(frase)
print("🔹 Análisis con spaCy:")
for token in doc:
    print(f"   {token.text} → {token.pos_}")
```
