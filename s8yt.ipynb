```python
# Importamos spaCy, una librería de NLP que nos permite trabajar
# con texto utilizando modelos lingüísticos y diferentes herramientas.
import spacy

# Importamos 'dictionary' de SudachiPy, un tokenizador especialmente
# útil para trabajar con japonés.
from sudachipy import dictionary

# Importamos jieba, una librería especializada en segmentar
# texto en chino.
import jieba

# Importamos AutoTokenizer de Transformers.
# Nos permite cargar tokenizadores utilizados por modelos
# de lenguaje como BERT, RoBERTa, GPT, etc.
from transformers import AutoTokenizer

# Importamos Stanza, una librería de NLP desarrollada por Stanford
# que incluye modelos para muchos idiomas.
import stanza

# Si todas las importaciones funcionan, significa que las librerías
# están instaladas y Python puede encontrarlas correctamente.
print("Todo instalado correctamente")
```

```python
# ─────────────────────────────────────────────
# ESPAÑOL E INGLÉS → spaCy
# ─────────────────────────────────────────────

# Cargamos el modelo pequeño de spaCy para español.
# Este modelo contiene información lingüística específica del idioma
# y nos permitirá procesar y tokenizar textos en español.
nlp_es = spacy.load("es_core_news_sm")

# Hacemos lo mismo con inglés, utilizando su modelo específico.
# Es importante usar un modelo adaptado al idioma porque
# las reglas de tokenización pueden variar entre lenguas.
nlp_en = spacy.load("en_core_web_sm")

# ─────────────────────────────────────────────
# JAPONÉS → SudachiPy
# ─────────────────────────────────────────────

# Creamos un diccionario de SudachiPy y, a partir de él,
# construimos nuestro tokenizador para japonés.
#
# SudachiPy está diseñado específicamente para analizar
# y segmentar texto japonés.
tokenizer_ja = dictionary.Dictionary().create()

# ─────────────────────────────────────────────
# FINÉS → spaCy
# ─────────────────────────────────────────────

# Cargamos el modelo de spaCy específico para finés.
# Así podemos comparar cómo funciona la tokenización
# en una lengua con características muy diferentes
# del español y del inglés.
nlp_fi = spacy.load("fi_core_news_sm")

# ─────────────────────────────────────────────
# ÁRABE → Stanza
# ─────────────────────────────────────────────

# Descargamos los recursos lingüísticos necesarios
# para procesar árabe con Stanza.
stanza.download("ar")

# Creamos un pipeline de Stanza para árabe.
# Indicamos que, en este caso, solo necesitamos el procesador
# de tokenización, porque nuestro objetivo es segmentar el texto.
nlp_ar = stanza.Pipeline(
    "ar",
    processors="tokenize"
)
```

```python
texto_es = (
    "Para alcanzar el objetivo, lo mejor es adoptar un enfoque práctico-teórico; p. ej., ¡hacer esto funcionará!"
)

doc_es = nlp_es(texto_es)

tokens_es = [
    token.text
    for token in doc_es
]

num_tokens_es = len(tokens_es)

print(tokens_es)
```

```python
# Creamos un texto de ejemplo en español.
# Hemos elegido una frase con diferentes elementos que pueden
# resultar interesantes para la tokenización:
# palabras, un guion, un punto abreviado, signos de puntuación
# y una exclamación.
texto_es = (
    "Para alcanzar el objetivo, lo mejor es adoptar un enfoque práctico-teórico; p. ej., ¡hacer esto funcionará!"
)

# Pasamos el texto por el modelo de spaCy para español.
# spaCy analiza la frase y determina cómo debe dividirla en tokens.
doc_es = nlp_es(texto_es)

# Recorremos todos los elementos que spaCy ha identificado
# y guardamos únicamente su texto.
#
# token.text contiene exactamente la forma que aparece
# en el texto original.
tokens_es = [
    token.text
    for token in doc_es
]

# Contamos cuántos tokens ha generado spaCy.
# len() nos devuelve el número de elementos de la lista.
num_tokens_es = len(tokens_es)

# Mostramos en pantalla la lista final de tokens.
print(tokens_es)
```

```python
# Creamos el tokenizador de SudachiPy para japonés.
# SudachiPy está diseñado para segmentar texto japonés,
# donde los espacios no suelen marcar los límites entre palabras.
tokenizer_ja = dictionary.Dictionary().create()

# Creamos un texto de ejemplo en japonés.
# La frase contiene puntuación y palabras que pueden segmentarse
# de diferentes maneras dependiendo del criterio utilizado.
texto_ja = (
    "目標を達成するには、実践的・理論的なアプローチを採用するのが最善です。例えば、これを行えばうまくいきます！"
)

# Tokenizamos el texto utilizando SudachiPy.
tokens_ja = [
    # surface() devuelve la forma exacta del texto que aparece
    # en cada token.
    m.surface()
    
    # tokenize() divide el texto siguiendo las reglas lingüísticas
    # de SudachiPy.
    for m in tokenizer_ja.tokenize(
        
        # Le pasamos nuestro texto en japonés.
        texto_ja,
        
        # SplitMode.C utiliza uno de los niveles de segmentación
        # disponibles en SudachiPy.
        # El modo C realiza una segmentación más amplia,
        # agrupando unidades que pueden considerarse palabras
        # o unidades léxicas más grandes.
        tokenizer_ja.SplitMode.C
    )
]

# Contamos cuántos tokens hemos obtenido.
num_tokens_ja = len(tokens_ja)

# Mostramos la lista de tokens resultante.
print(tokens_ja)
```

```python
# Creamos un texto de ejemplo en chino.
# Al igual que ocurre con el japonés, el chino normalmente no utiliza
# espacios para separar las palabras, por lo que necesitamos
# un tokenizador que determine dónde están los límites.
texto_zh = (
    "为了达到目标，最好采用实践与理论相结合的方法；例如，这样做会成功！"
)

# Pasamos el texto por jieba para segmentarlo.
# jieba.cut() analiza la secuencia de caracteres y propone
# una segmentación en unidades que podemos tratar como tokens.
tokens_zh = list(
    jieba.cut(texto_zh)
)

# Contamos cuántos tokens ha generado jieba.
num_tokens_zh = len(tokens_zh)

# Mostramos la lista de tokens obtenida.
print(tokens_zh)
```

```python
# Creamos un texto de ejemplo en finés.
# Incluimos un compuesto con guion, puntuación y una exclamación
# para observar cómo spaCy identifica los diferentes tokens.
texto_fi = (
    "Tavoitteen saavuttamiseksi on parasta omaksua käytännöllis-teoreettinen lähestymistapa; esimerkiksi tämän tekeminen toimii!"
)

# Pasamos el texto por el modelo de spaCy para finés.
# El modelo aplica las reglas de tokenización específicas del idioma.
doc_fi = nlp_fi(texto_fi)

# Recorremos los elementos que spaCy ha identificado como tokens
# y guardamos su representación textual.
tokens_fi = [
    token.text
    for token in doc_fi
]

# Mostramos la lista de tokens resultante.
print(tokens_fi)
```

```python
# Cargamos un tokenizador de Hugging Face Transformers.
# En este caso utilizamos mBERT (BERT multilingüe), un modelo
# entrenado con texto de muchos idiomas, incluido el finés.
tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-multilingual-cased"
)

# Tokenizamos el texto en finés utilizando el tokenizador de mBERT.
# A diferencia de spaCy, aquí trabajamos con subwords:
# una palabra puede dividirse en varias unidades más pequeñas.
tokens_fi_subword = tokenizer.tokenize(texto_fi)

# Contamos cuántos subwords hemos obtenido.
num_tokens_fi_subword = len(tokens_fi_subword)

# Mostramos la lista de subwords resultante.
print(tokens_fi_subword)
```

```python
# Creamos un texto de ejemplo en árabe.
# Incluimos puntuación y un compuesto con guion para observar
# cómo Stanza segmenta diferentes elementos del texto.
texto_ar = (
    "لتحقيق الهدف، من الأفضل اعتماد نهج عملي-نظري؛ على سبيل المثال، القيام بذلك سينجح!"
)

# Pasamos el texto por el pipeline de Stanza para árabe.
# Como configuramos el pipeline con el procesador "tokenize",
# Stanza se encarga de identificar los tokens del texto.
doc_ar = nlp_ar(texto_ar)

# Recorremos las frases que Stanza ha detectado en el documento
# y, dentro de cada frase, recorremos sus palabras.
#
# word.text contiene la forma original de cada token.
tokens_ar = [
    word.text
    for sentence in doc_ar.sentences
    for word in sentence.words
]

# Contamos cuántos tokens hemos obtenido.
num_tokens_ar = len(tokens_ar)

# Mostramos la lista final de tokens.
print(tokens_ar)
```

```python
# Creamos una función para experimentar qué ocurre cuando utilizamos
# el tokenizador de inglés sobre textos de otros idiomas.
def experimento_tokenizador_ingles():
    
    # Creamos una lista con los idiomas que queremos poner a prueba.
    # Para cada idioma guardamos tres elementos:
    # 1. El nombre del idioma.
    # 2. El texto original.
    # 3. Los tokens obtenidos con el tokenizador que hemos elegido
    #    como referencia para ese idioma.
    casos_test = [
        ("Español", texto_es, tokens_es),
        ("Japonés", texto_ja, tokens_ja),
        ("Chino", texto_zh, tokens_zh),
        ("Árabe", texto_ar, tokens_ar),
        ("Finés", texto_fi, tokens_fi_subword),
    ]

    # Recorremos cada uno de los casos de prueba.
    for idioma, texto, tokens_correctos in casos_test:

        # Pasamos el texto por el modelo de spaCy para inglés.
        # Aquí está el experimento:
        # utilizamos un tokenizador pensado para inglés
        # sobre textos escritos en otros idiomas.
        tokens_ingles = [
            token.text
            for token in nlp_en(texto)
        ]

        # Mostramos el nombre del idioma que estamos probando.
        print(f"\n🌍 \033[1m{idioma}\033[0m")

        # Mostramos la frase original sin modificar.
        print(f"📝 \033[4mFrase original\033[0m: {texto}")

        # Mostramos cómo ha segmentado el texto el tokenizador de inglés.
        print(f"❌ \033[4mTokenizador inglés\033[0m: {tokens_ingles}")

        # Mostramos la segmentación obtenida con el tokenizador
        # que hemos elegido como referencia para ese idioma.
        print(f"✅ \033[4mTokenizador {idioma.lower()}\033[0m: {tokens_correctos}")

        # Comparamos cuántos tokens ha producido cada sistema.
        # Esto nos permite detectar diferencias en la segmentación,
        # aunque el número de tokens por sí solo NO determina
        # cuál es una tokenización "correcta".
        print(
            f"📊 \033[4mTokens\033[0m: "
            f"{len(tokens_ingles)} tokens en tokenizador inglés "
            f"vs {len(tokens_correctos)} tokens en tokenizador {idioma.lower()}."
        )

# Ejecutamos la función para realizar todo el experimento.
experimento_tokenizador_ingles()
```
