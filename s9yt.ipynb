```python
# Importamos CountVectorizer de scikit-learn.
# Esta herramienta convierte textos en una representación numérica
# basada en la frecuencia de las palabras.
from sklearn.feature_extraction.text import CountVectorizer

# Creamos una lista con nuestros dos textos de ejemplo.
# Cada elemento de la lista es un documento independiente.
textos = [
    "El gato negro persigue al ratón gris",
    "El perro marrón duerme en el jardín"
]

# Creamos el vectorizador.
# Todavía no hemos procesado ningún texto:
# simplemente estamos creando el objeto que hará la transformación.
vectorizer = CountVectorizer()

# Aprendemos el vocabulario a partir de nuestros textos
# y los transformamos en una matriz de frecuencias.
X = vectorizer.fit_transform(textos)

# Obtenemos las palabras que forman el vocabulario.
# Cada palabra se convierte en una columna de nuestra matriz.
print("Vocabulario:", vectorizer.get_feature_names_out())

# Convertimos la matriz de scikit-learn a un array de NumPy
# para poder visualizarla de forma sencilla.
# Cada fila representa un documento
# y cada columna representa una palabra del vocabulario.
# El número indica cuántas veces aparece esa palabra en ese documento.
print("Frecuencias:\n", X.toarray())
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Dos nuevos textos de ejemplo para repetir el mismo proceso
textos_2 = ["El perro muerde al hombre", 
          "El hombre muerde al perro"]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(textos_2)

print("Vocabulario:", vectorizer.get_feature_names_out())
print("Frecuencias:\n", X.toarray())
```

```python
# Importamos CountVectorizer para convertir los textos
# en una representación numérica basada en palabras y n-gramas.
from sklearn.feature_extraction.text import CountVectorizer

# Con ngram_range=(1, 2) le decimos al vectorizador
# que queremos capturar n-gramas de tamaño 1 y 2:
# palabras individuales (unigramas) y pares de palabras consecutivas (bigramas).
vectorizer = CountVectorizer(ngram_range=(1, 2))

# Creamos dos textos de ejemplo.
# En este caso, nos interesa especialmente observar
# las combinaciones de palabras que aparecen en cada frase.
textos = [
    "La película no es buena",
    "La película es muy buena"
]

# El vectorizador aprende el vocabulario de los textos
# y los convierte en una matriz numérica.
X = vectorizer.fit_transform(textos)

# Recuperamos todos los n-gramas que ha aprendido el vectorizador.
# Aquí aparecerán tanto los unigramas como los bigramas.
print("Unigramas y bigramas:", vectorizer.get_feature_names_out())

# También podemos separar los unigramas de los bigramas.
# Los bigramas contienen un espacio porque están formados por dos palabras.
unigramas = [p for p in vectorizer.get_feature_names_out() if " " not in p]
bigramas = [p for p in vectorizer.get_feature_names_out() if " " in p]

print("Unigramas:", unigramas)
print("Bigramas:", bigramas)
```

```python
# Importamos TfidfVectorizer de scikit-learn.
# Esta herramienta transforma los textos en valores numéricos
# asignando un peso a cada palabra según su importancia.
from sklearn.feature_extraction.text import TfidfVectorizer

# Creamos nuestro vectorizador TF-IDF.
vectorizer = TfidfVectorizer()

# Analizamos los textos y los transformamos
# en una matriz de valores TF-IDF.
X = vectorizer.fit_transform(textos)

# Importamos pandas para poder visualizar
# los resultados en forma de tabla.
import pandas as pd

# Convertimos la matriz TF-IDF en un DataFrame.
# Cada fila representa un documento.
# Cada columna representa una palabra.
# Cada valor es el peso TF-IDF de esa palabra en ese documento.
tfidf_df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

# Mostramos la tabla con los pesos TF-IDF.
print(tfidf_df)
```
