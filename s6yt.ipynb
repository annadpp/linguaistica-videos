```python
# CONDICIONALES 2

def clasificar_figura(frase):
    frase = frase.lower()
    
    if " como " in frase or " cual " in frase or " igual que " in frase:
        return "símil"
    elif any(palabra in frase for palabra in [" es ", " son ", " eran ", " fueron "]):
        return "metáfora"
    elif "parecido a" in frase or "semejante a" in frase:
        return "comparación"
    else:
        return "literal"


# Y ahora, análisis automático:
frases = [
    "El tiempo es un río que fluye",
    "El amor como una rosa",
    "Sus ojos eran estrellas",
    "Corría veloz como el viento",
    "La casa estaba silenciosa"
]


for frase in frases:
    clasificacion = clasificar_figura(frase)
    print(f"'{frase}' → {clasificacion}")
```

```python
# VIAJE FINAL 2 (la revolución)

import spacy
from nltk.corpus import gutenberg
import nltk

# Descargar datos la primera vez
nltk.download('gutenberg')

# Cargar modelo de spaCy (en inglés, porque los textos son en inglés)
nlp = spacy.load("en_core_web_sm")  # asegúrate de tenerlo instalado: !python -m spacy download en_core_web_sm

# 1. Defino autores y textos
autores = {
    "Austen": ["austen-emma.txt"],
    "Shakespeare": ["shakespeare-hamlet.txt"],
    "Milton": ["milton-paradise.txt"]
}

# 2. Patrones que vamos a contar (en inglés)
patrones = ["love", "death", "time", "nature", "life"]

# 3. Función de análisis
def analizar_autor(archivos, autor):
    resultados_autor = []
    
    for archivo in archivos:
        # Cargar texto completo del corpus
        texto = gutenberg.raw(archivo)
        doc = nlp(texto)
        
        # Tokens alfabéticos
        tokens_validos = [t.text.lower() for t in doc if t.is_alpha]
        
        # Riqueza léxica
        riqueza = len(set(tokens_validos)) / len(tokens_validos) if tokens_validos else 0
        
        # Conteo de patrones
        conteo_patrones = {patron: tokens_validos.count(patron) for patron in patrones}
        
        resultados_autor.append({
            "archivo": archivo,
            "riqueza_lexica": riqueza,
            "patrones": conteo_patrones
        })
    
    return resultados_autor

# 4. Ejecución del análisis
print("📊 ANÁLISIS COMPARATIVO GUTENBERG")
print("=" * 60)

resultados_totales = {}

for autor, archivos in autores.items():
    print(f"\n🎭 Analizando {autor}...")
    resultados = analizar_autor(archivos, autor)
    resultados_totales[autor] = resultados
    
    # Riqueza promedio
    riqueza_promedio = sum(r["riqueza_lexica"] for r in resultados) / len(resultados)
    print(f"   📈 Riqueza léxica promedio: {riqueza_promedio:.3f}")
    
    for patron in patrones:
        total_patron = sum(r["patrones"][patron] for r in resultados)
        print(f"   🔍 '{patron}': {total_patron} apariciones")

print("\n" + "=" * 60)
print("✅ ANÁLISIS COMPLETADO: 3 textos, 5 patrones cada uno")
```
